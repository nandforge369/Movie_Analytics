```python
movies["release_date"] = pd.to_datetime(movies["release_date"].str.strip(),errors="coerce")
movies["Year_release"] = movies["release_date"].dt.year 
```

```python
x=movies.groupby("Year_release").size().sort_values(ascending=False).head(1)
x
```

```python
Q3 = movies["runtime"].quantile(0.75)
Q1 = movies["runtime"].quantile(0.25)
IQR = Q3-Q1
LL = Q1-1.5*IQR
UL = Q3+1.5*IQR

movies[(movies["runtime"]>UL) | (movies["runtime"]<LL)]["title"]
```

```python
# 1. Sort genres so the first listed genre comes first for each movie
movie_genres_sorted = movie_genres.sort_values(by=["movie_id", "genre_id"])

# 2. Select the first genre for each movie
primary_genres = movie_genres_sorted.drop_duplicates(subset="movie_id",keep="first")

# 3. Add genre names
primary_genres = primary_genres.merge(genres,on="genre_id",how="left")

# 4. Add movie popularity
movie_primary_genre = movies.merge(primary_genres[["movie_id", "genre_name"]],on="movie_id",how="left")

# 5. Calculate average popularity by primary genre
avg_popularity = (
    movie_primary_genre
    .groupby("genre_name")["popularity"]
    .mean()
    .sort_values(ascending=False)
    .reset_index(name="avg_popularity")
)

print(avg_popularity)
```

```python
prolific_10 = (cast.groupby("actor_name")["movie_id"].nunique().sort_values(ascending=False).head(10).reset_index(name="movie_count"))
top_cast = cast.merge(prolific_10[["actor_name"]],on="actor_name",how="inner")
top_cast = top_cast.merge(movies[["movie_id", "vote_average"]],on="movie_id",how="left")
actor_ratings = (top_cast.groupby("actor_name").agg(movie_count=("movie_id", "nunique"),avg_rating=("vote_average", "mean")).reset_index())
result = (actor_ratings.sort_values(by=["movie_count", "avg_rating"],ascending=[False, False]).head(10))
print(result)
```
